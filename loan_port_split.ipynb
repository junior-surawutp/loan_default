# Train-Test Splitting

```python
loan_port2=loan_port.copy()

datetime_col=[i for i in loan_port.keys() if 'date' in i]
# transform str to datetime
for i in datetime_col: 
    loan_port2[i]=pd.to_datetime(loan_port2[i])

loan_port2
```

```python
loan_port2.sort_values(['origination_date','maturity_date'], inplace=True)
loan_port2.reset_index(drop=True, inplace=True)
```

```python
split_point=int(len(loan_port2)*0.7)

train_data=loan_port2[:split_point].copy()
test_data=loan_port2[split_point:].copy()
```

```python
train_stat=train_data.describe()
display(train_stat)
```

```python
# prepare for forward transform

month_col=[i for i in loan_port2.keys() if 'month' in i]
dummy_col=['sector','loan_type','collateral']
z_scale_col=['ead', 'coupon_rate', 'leverage', 'interest_coverage', 'debt_to_equity', 'pd_annual', 'lgd', 'el', 'unexpected_loss', 'rwa', 'recovery_rate', 'loss_given_default']
#'pd_annual', 'lgd' and 'recovery_rate' already in interval 0-1. 'defaulted' is boolean

# re-scale initial rating
def rate_to_num(rate_char):
    if rate_char=='AAA': return 7/7
    elif  rate_char=='AA': return 6/7
    elif  rate_char=='A': return 5/7
    elif  rate_char=='BBB': return 4/7
    elif  rate_char=='BB': return 3/7
    elif  rate_char=='B': return 2/7
    elif  rate_char=='CCC': return 1/7
    else: return 0

# transform initial rating as number 0-1 to string
def num_to_rate(num):
    temp=np.round(num*7)
    if temp==7: return'AAA'
    elif temp==6: return 'AA'
    elif temp==5: return 'A'
    elif temp==4: return 'BBB'
    elif temp==3: return 'BB'
    elif temp==2: return 'B'
    elif temp==1: return 'CCC'
    else: return 'D'
```

```python
display(month_col)
display(datetime_col)
```

```python
def forward_transform(loan_data):
    loan_port2=loan_data.copy()

    # re-scale month to 0-1
    for i in month_col: 
        loan_port2[i]=(loan_port2[i]-12)/(120-12)

    # re-scale initial rating to 0-1
    loan_port2['initial_rating']=loan_port2['initial_rating'].apply(rate_to_num) 

    # re-scale credit score to 0-1
    loan_port2['credit_score']=(loan_port2['credit_score']-300)/(850-300)
    
    # make one-hot encoder
    dummy_get=[pd.get_dummies(loan_port2[i]).astype('int') for i in dummy_col]
    dummy_col2=sum([list(dummy_get[i].keys()) for i in range(len(dummy_col))],[])
    for i in range(len(dummy_col)):
        loan_port2=pd.merge(loan_port2, dummy_get[i], left_index=True, right_index=True)

    # transform to z-score
    for i in z_scale_col:
        loan_port2[i]=(loan_port2[i]-train_stat.loc['mean',i])/train_stat.loc['std',i]

    return loan_port2
```

```python
train_data2=forward_transform(train_data)
train_data2
```

```python
def backward_transform(loan_data):
    loan_port2=loan_data.copy()

    # re-scale month
    for i in month_col: 
        loan_port2[i]=(loan_port2[i]*(120-12)+12).astype('int')

    # re-scale initial rating to 0-1
    loan_port2['initial_rating']=loan_port2['initial_rating'].apply(num_to_rate) 

    # re-scale credit score
    loan_port2['credit_score']=(loan_port2['credit_score']*(850-300)+300).astype('int')
    
    # transform one-hot encoder into columns
    dummy_backward2=[pd.from_dummies(loan_port2[dummy_backward[i]]).rename(columns={'':dummy_col[i]}) for i in range(len(dummy_col)) ]
    for i in range(len(dummy_col)):
        loan_port2=pd.merge(loan_port2, dummy_backward2[i], left_index=True, right_index=True, suffixes=(None,'_f'))

    # transform to z-score
    for i in z_scale_col:
        loan_port2[i]=loan_port2[i]*train_stat.loc['std',i]+train_stat.loc['mean',i]

    return loan_port2
```
